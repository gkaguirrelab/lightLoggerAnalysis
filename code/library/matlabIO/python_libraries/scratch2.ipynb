```python
import numpy as np
import pandas as pd

importlib.reload(chunk_util)

raw_floor_threshold: int = 16
world_chunk_paths: list[tuple[str, str]] = chunk_util.group_sensors_files(path_to_raw)["W"]

low_raw_value_rows: list[dict] = []
global_frame_idx: int = 0

for chunk_idx, chunk_paths in enumerate(world_chunk_paths):
    print(f"Inspecting raw world chunk {chunk_idx + 1} / {len(world_chunk_paths)}")

    # Parse one raw world chunk at a time. Keep every transform flag off so these
    # are the values before DGain, debayering, fielding, linearization, or video IO.
    parsed_chunk = chunk_util.world_chunk_parser(
        chunk_paths,
        apply_digital_gain=False,
        convert_time_units=False,
        use_mean_frame=False,
        convert_to_float=False,
        apply_phase_offset=False,
        apply_RGB_correction=False,
        apply_fielding_function=False,
        debayer_images=False,
        contains_AGC_metadata=True,
        linearize_camera_responsivitiy=False,
        apply_floor_ceiling=False,
    )
    raw_frames = parsed_chunk["W"]["v"]
    raw_timestamps = parsed_chunk["W"]["t"]

    for frame_idx_in_chunk, raw_frame in enumerate(raw_frames):
        zero_mask = raw_frame == 0
        below_threshold_mask = raw_frame < raw_floor_threshold
        if(np.any(zero_mask) or np.any(below_threshold_mask)):
            first_locations = np.argwhere(zero_mask | below_threshold_mask)[:10]
            low_raw_value_rows.append({
                "chunk_idx": chunk_idx,
                "frame_idx_in_chunk": frame_idx_in_chunk,
                "global_frame_idx": global_frame_idx,
                "timestamp_ns": raw_timestamps[frame_idx_in_chunk],
                "frame_dtype": str(raw_frame.dtype),
                "frame_min": np.min(raw_frame),
                "frame_max": np.max(raw_frame),
                "zero_pixel_count": int(np.count_nonzero(zero_mask)),
                f"pixel_count_below_{raw_floor_threshold}": int(np.count_nonzero(below_threshold_mask)),
                "first_low_pixel_locations_row_col": first_locations.tolist(),
            })

        global_frame_idx += 1

    del parsed_chunk, raw_frames

low_raw_value_df = pd.DataFrame(low_raw_value_rows)
display(low_raw_value_df)

if(len(low_raw_value_df) == 0):
    print(f"No raw world pixels equal to 0 or below {raw_floor_threshold} were found.")
else:
    print(f"Found {len(low_raw_value_df)} frames with raw world pixels equal to 0 or below {raw_floor_threshold}.")
```
